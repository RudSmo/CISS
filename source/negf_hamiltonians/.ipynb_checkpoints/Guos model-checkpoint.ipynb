```julia
function H_Guo_dna(
        onsight_c1::Array{Array},
        onsight_c2::Array{Array},
        thopping::Array{Float64}, 
        lambda::Array{Float64},
        )
    N = length(onsight_c1[1])* 4
    onsight_e = [onsight_c1; onsight_c2]
    
    Mtrx = zeros(Float64, N,N)
    for i in 1:4:N # i = number of site n bloc
        i_steps = Integer((i+3)/4)
        Mtrx[i+2, i] = lambda[i_steps] 
        Mtrx[i, i+2] = lambda[i_steps]
        Mtrx[i+3, i+ 1] = lambda[i_steps] 
        Mtrx[i+ 1, i+3] = lambda[i_steps]
        
        for j in 1:1:4 # j = indexing  inside the bloc
        
        #diagonaleinträge
        Mtrx[(i+ j -1), (i+ j - 1)] = onsight_e[j][i_steps]
        
            if i<N-4
                #rechte hopping matrix
                Mtrx[(i +3 +j), (j +i-1)] = thopping[j]  
                #untere hoping matrix
                Mtrx[(j +i -1), (i+3 + j)] = thopping[j]
            end   
        end
    end
    return Matrix(Mtrx)
end 

function H_Guo_dna2(onsight_spin_up::Array{Float64},
        onsight_spin_down::Array{Float64},
        thopping::Array{Float64}, 
        lambda::Array{Float64},
        N::Integer)
    onsight_e = Array[onsight_spin_up, onsight_spin_down] # each have to have length N
    
    Mtrx = zeros(Float64, 2*N,2*N)
    for i in 1:4:2*N # i = position of site n bloc in grand matrix
        i_steps = Integer((i+3)/4) # number of a site. Is one number per i
        print(i_steps)
        Mtrx[i+2, i] = lambda[i_steps] # lambda is interstrand (inside one strand) hopping
        Mtrx[i, i+2] = lambda[i_steps]
        Mtrx[i+3, i+ 1] = lambda[i_steps] # j = 2
        Mtrx[i+ 1, i+3] = lambda[i_steps] # j = 2

        for spin in 1:1:2 # spin = 1 = up and spin = 2 = down 

        #diagonaleinträge
        Mtrx[(i+ spin -1), (i+ spin - 1)] = onsight_e[spin][i_steps] #  j = 1
        Mtrx[(i+ spin +1), (i+ spin +1)] = onsight_e[spin][Integer(N / 2)  + i_steps] # j = 2
        # ------------------------------------------------use N/2 since N is the length of aray spin up or down 
        # for [j = 1, j=2] after half length of that array we switch to the other strand !
            
            if i<2*N-4# t is intrastrand hopping ( between strands)
            #rechte hopping matrix
            Mtrx[(i+3+spin), (spin+i-1)] = thopping[spin] # j = 1
            Mtrx[(i+5+spin), (spin+i+1)] = thopping[spin + 2] # j = 2

            #untere hoping matrix
            Mtrx[ (spin+i-1), (i+3+spin)] = thopping[spin] # j = 1
            Mtrx[ (spin+i+1), (i+5+spin)] = thopping[spin + 2] # j = 2

            end   
        end
       
    end#???????????????????check if thopping is ordered like this?!!
    return Matrix(Mtrx)
end
```

```julia
function sigma_ort(phi::Float64, theta::Float64)
    sig_x = Matrix([0 1; 1 0])
    sig_y = Matrix([0 -1im; 1im 0 ])
    sig_z = Matrix([1 0; 0 -1])
    return sig_x * sin(phi)*sin(theta) - sig_y* cos(phi)*sin(theta) + sig_z* cos(theta)
end
function sites_to_angle(sites_per_lap::Integer)# number of sites per helical turn
    Radius = 0.7* 10^(-9)
    delta_phi = pi* 2/sites_per_lap
    l_a = Radius/sites_per_lap
    return [delta_phi l_a]
end   
function H_Guo_so(
        onsight_c1::Array{Array},
        onsight_c2::Array{Array},
        sites_per_turn::Integer, 
        theta::Float64
        )
    dim = length(onsight_c1[1])* 4
    Mtrx = zeros(Complex{Float64}, dim,dim)  
    
    # ? natürliche Einheiten oder SI???
    #h_quer = 1
   # R = 0.7e-9
    #dV_dr_von_R = 1
    #m = 1
    #c =1
    alpha = 1 # h_quer^2/(4*m^2*c^2)* dV_dr_von_R
    l_a = sites_to_angle(sites_per_turn)[2]
    t_so = -alpha/(4*l_a)
    delta_phi = sites_to_angle(sites_per_turn)[1]
   
    for i in 1:4:dim  # i = position of site n in full matrix, i_steps =  number of site n
        i_steps = Integer((i+3)/4)
        if i<dim-4
            for z in 0:2:2 
                # für z  = 0 git j = 1 und z = 2 ist j = 2 
                if z ==0
                    sig_n_und_npl1 = t_so*1im* (sigma_ort( i_steps*delta_phi, theta) .+ sigma_ort(( i_steps -1 ) *delta_phi, theta) )
                    
                    #adresses a 2x2 block matrix #rechte hopping matrix h.c. seite
                    Mtrx[(i+z):(i+z+1), (i+4+z):(i+5+z)] = sig_n_und_npl1'
                    #untere hopping matrix -> ohne compl conj?! 
                    Mtrx[(i+z+4):(i+z+5), (i+z):(i+1+z)] = sig_n_und_npl1 
                end
                if z == 2  #j = helical chain = 2 : plus pi in both sigma functions. As pi is half a circle adding this to phi results being on the other helical chain
                    sig_n_und_npl1_pi =t_so* 1im* (sigma_ort( i_steps*delta_phi + pi, theta) .+ sigma_ort(( i_steps -1 ) *delta_phi + pi, theta))
                    Mtrx[(i+z):(i+z+1), (i+4+z):(i+5+z)] = sig_n_und_npl1_pi'
                    Mtrx[(i+z+4):(i+z+5), (i+z):(i+1+z)] =sig_n_und_npl1_pi
                end
            end
        end   
    end
    return Matrix(Mtrx) # !! complex matrix
end     
function H_Guo_so_sort2(
        sites_per_lap::Integer, 
        theta::Float64, 
        N::Int64 # total number of sites in both chains
        )
    Mtrx = zeros(Complex{Float64}, 2*N,2*N)# since we have two states: up and down for each site  
   
    alpha = 1
    l_a = sites_to_angle(sites_per_lap)[2]
    t_so = 1#-alpha/(4*l_a)
    delta_phi = sites_to_angle(sites_per_lap)[1]
    
    # i = position of site n in full matrix, i_steps =  number of site n
    for i in 1:4:2*N 
        i_steps = Integer((i+3)/4)
        if i<2*N-4
            for z in 0:2:2 
                # für z  = 0 git j = 1 und z = 2 ist j = 2 
                if z ==0
                    #rechte hopping matrix
                    #* ist dies die h.c. seite oder die unteren matrizen?
                    sig_n_und_npl1 = 1im* (sigma_ort( i_steps*delta_phi, theta) .+ sigma_ort(( i_steps -1 ) *delta_phi, theta) )
                   #adresses a 2x2 block matrix
                    Mtrx[(i+z):(i+z+1), (i+4+z):(i+5+z)] = t_so*sig_n_und_npl1'
                    
                    #untere hopping matrix -> ohne compl conj?! 
                   Mtrx[(i+z+4):(i+z+5), (i+z):(i+1+z)] = t_so*(sig_n_und_npl1 )
                   
                end
                if z == 2
                    #plus pi in both sigma functions. As pi is half a circle adding this to phi results being on the other helical chain
                    sig_n_und_npl1 = 1im* (sigma_ort( i_steps*delta_phi + pi, theta) .+ sigma_ort(( i_steps -1 ) *delta_phi + pi, theta))
                    Mtrx[(i+z):(i+z+1), (i+4+z):(i+5+z)] = t_so*sig_n_und_npl1'
                    Mtrx[(i+z+4):(i+z+5), (i+z):(i+1+z)] = t_so*sig_n_und_npl1 
                end
            end
        end   
    end
    return Matrix(Mtrx)
end 
```

```julia
function H_Guo_elvib(
        g_couling::Array{Array}, 
        displacement::Array{Array}, # array with four arrays ordered: j = 1 up, j  = 1 down, j = 2 up, j = 2 down
        onsight_c1::Array{Array} 
    )
    dim = length(onsight_c1[1])* 4
    Mtrx = zeros(Complex{Float64}, dim,dim)
    
    for i in 1:4:dim # i = number of site n bloc * 4 -> position in grand matrix
        i_steps = Integer((i+3)/4)
        for j in 1:1:4 # j = indexing  inside the bloc
            #diagonaleinträge
            if j == 1 || j ==2
                # oder 2 displacement von 1, isteps,   else displacement von 2
                Mtrx[(i+ j -1), (i+ j - 1)] = displacement[1][i_steps] *g_coupling[j][i_steps]/ sqrt(2)
               
            else 
                Mtrx[(i+ j -1), (i+ j - 1)] = displacement[2][i_steps] *g_coupling[j][i_steps]/ sqrt(2)
             
            end
        end
    end
    return Mtrx
end
# is a imaginary matrix

function H_Guo_elvib_sort2(
        g_coupling::Float64, 
        X::Array{Float64}, # array with enumerating over n first half of array for j = 1, second half for j = 2 starting with n=0 again
        N::Int64 # number of total sites in both chains
    )
   
    Mtrx = zeros(Complex{Float64},2*N, 2*N)
    
    for i in 1:4:2*N # i = number of site n bloc * 4 -> position in grand matrix
        i_steps = Integer((i+3)/4)
        for k in 1:1:4 # k = indexing  inside the bloc
            #diagonaleinträge
            
            if k == 1 || k ==2 #|| heißt osder, da spin up und spin down hier keine unterschiede aufweisen
            #erster und zweiter Blockelement der Untermatriyx j = 1 --> x[i_ssteps]
                Mtrx[(i+k-1), (i+k-1)] = X[i_steps] *g_coupling/ sqrt(2)
            else # für j = 2 -> ab hälfte der x liste ---> x[N + i_steps]
                Mtrx[(i+k-1), (i+k-1)] = X[Integer(N/2) + i_steps]*g_coupling/ sqrt(2)
            end
        end
    end
    return Matrix(Mtrx)
end
```

```julia
function G_und_trans(energy::Float64,
        eta::Float64,
        Hamiltonian::Array{Complex{Float64}},
        gammas::Array{Complex{Float64}}
    )
    dim = size(Hamiltonian)[1]
    Matrix_I_e = (energy + eta*1im ) * Matrix(I, dim,dim)

    gamma_l = zeros(Complex{Float64}, dim,dim)
    gamma_l[1, 1] = gammas[1]
    gamma_l[2,2] = gammas[1]
    # as spin up and spin down sttate of first site are couled to lead? analogue for last of course
    #?!
    gamma_r = zeros(Complex{Float64}, dim,dim)
    gamma_r[end, end] = gammas[2]
    gamma_r[end-1, end-1] = gammas[2]
    
    matrix_self = -1im *(gamma_l+ gamma_r)/ 2 
    G_ret =  inv(Matrix_I_e - matrix_self - Hamiltonian)
    G_adv = adjoint(G_ret)
    final_Matrix = gamma_l* G_adv * gamma_r * G_ret
    return real.(tr(final_Matrix))
end 
function trans_plot(
        Hamiltonian::Array{Complex{Float64}},
        gammas::Array{Complex{Float64}},
        emin::Float64, 
        emax::Float64, 
        pdf_name::String)
    
    E_list = Array(range(emin, emax, length = 2000))
    eta = 10^(-7)
    transm_H = [G_und_trans(E,eta, H_all, gammas) for E in E_list]
    plt.figure()
    plt.plot(
        E_list, transm_H,    #mit f. normalerweise komponentenweise
        label = "transmission",
        linewidth = 2 
        )
    plt.legend()
    #plt.ylim( 0, 0.02)
    plt.title("Transmission H1 (degenerate energies for spin up and down) (n=5) at zero temp")
    plt.xlabel( "Energy in MeV")
    plt.ylabel( "Transmission coefficient")
    plt.grid("true")
    plt.savefig(pdf_name)
    plt.show()
    ei1, vec1 = eigen(H_all)
    print(ei1)
    return 
end
```

```julia
function integral_f(x_data::Array{Float64}, H::Array{Complex{Float64}}, 
        eta::Float64, gammas::Array{Complex{Float64}})
    y_int = []
    for i in 1:(length(x_data)-1)
        x_neu  = range(x_data[i], x_data[i+1], length = 10)
        
        y_neu = [G_und_trans(x/ 2,eta, H, gammas) for x in x_neu]
        if i ==1
            push!(y_int, trapz(x_neu, y_neu))
        end
        if i > 1
            s = y_int[i-1] +  trapz(x_neu, y_neu)
            push!(y_int, s)
        end
    end
        plt.plot(x_data[1:end-1], y_int, label= "current")
       
    return y_int
end
```
