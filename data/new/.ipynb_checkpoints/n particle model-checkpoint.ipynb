```julia
using LinearAlgebra
using PyPlot
using Trapz
using Calculus
```

```julia
function H_dna(
        energies,#::Array{Float64},
        thopping#::Float64
        )
    N = length(energies)
    M_all = zeros(Float64, N,N)
    for i in 1:N
        M_all[i, i] = energies[i]
        if i<N
            M_all[i, i+1] = thopping            
        end 
        if i > 1 
            M_all[i+1, i] = thopping
        end 
    end
    return M_all
end 
example = range(1,12, length =11)
print(H_dna(collect(example),0.3 ))
```
